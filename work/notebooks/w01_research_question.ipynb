## 2) The question, decision, action, and cost of a wrong call

**Search question:** Which pages should be reviewed first because they are losing search visibility while still having enough demand to matter?

**Unit of analysis:** One page (one content item).

**Output:** A ranked list of pages to review, with a score and a short reason for each.

**Decision it improves:** Which pages a reviewer should spend limited time on first.

**Action someone could take:** Review the top 20 to 50 pages and decide whether to refresh them, monitor them, or leave them alone.

**Cost of a wrong recommendation:** Reviewer time is wasted on healthy pages, or a page that keeps declining is missed and loses more traffic.

**Why data or ML can help:** A fixed rule treats every page the same way. Data lets us check whether a learned ranking finds more truly declining pages in its top results than a simple rule does.

**Why this is not just "train a model":** The goal is a ranked review list for humans, checked against a simple baseline. The model has to beat the rule to be worth using, and it only supports decisions. It does not prove that a refresh fixes a page.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/MahamFatima5/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Keep pages that have impressions and are at least 90 days old, one row per page
d = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]
d = d.drop_duplicates("content_id")

print("Pages in the starter file:", len(df))
print("Pages kept after filters:", len(d))
print("Share of pages trending down:", round((d["trend_direction"] == "down").mean() * 100, 1), "%")
print("Median impressions (90 days):", d["impressions_90d"].median())
print("Median content age (days):", d["content_age_days"].median())
```

**What the numbers show:** After keeping pages with impressions and at least 90 days of age, 30,000 pages remain. About 54.2% of them are trending down, and the median page has about 731 impressions in 90 days. This shows enough pages and enough decline to make a review ranking worthwhile.
